```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("inflation.db")
conn.close()
```

```python
with sqlite3.connect("inflation.db") as conn:
    conn.execute("DROP TABLE IF EXISTS cpi_base")
```

```python
with sqlite3.connect("inflation.db") as conn:
    conn.execute("""
        CREATE TABLE IF NOT EXISTS cpi_base (
            Date TEXT PRIMARY KEY,
            CPI REAL,
            CPI_annual_pct_change REAL,
            Updated TEXT
        )
    """)
```

```python
with sqlite3.connect("inflation.db") as conn:
    existing = pd.read_sql("SELECT * FROM cpi_base", conn)
existing
```
